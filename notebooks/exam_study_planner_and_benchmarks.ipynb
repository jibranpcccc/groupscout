This open Python study planner generates preparation calendars and peer benchmark metrics for the **GRE, GMAT, CFA, CPA, NEET, and JEE** examinations.

```python
# Diagnostic Study Milestone Planner
def generate_study_milestones(exam, total_study_hours, weekly_commitment):
    total_weeks = total_study_hours / weekly_commitment
    phase1_weeks = round(total_weeks * 0.4, 1)  # Core Theory & Syllabus
    phase2_weeks = round(total_weeks * 0.35, 1) # Question Bank & Peer Problem Solving
    phase3_weeks = round(total_weeks * 0.25, 1) # Full-length Timed Mocks
    
    return {
        "Exam": exam,
        "Total Weeks": round(total_weeks, 1),
        "Phase 1 (Core Concepts)": f"{phase1_weeks} weeks",
        "Phase 2 (Question Bank Drills)": f"{phase2_weeks} weeks",
        "Phase 3 (Mocks & Review)": f"{phase3_weeks} weeks",
        "Verified Peer Community": f"https://studygroupshub.com/exam/{exam.lower()}/"
    }

# Example: GRE Preparation (300 hours)
plan = generate_study_milestones("GRE", 300, 20)
for k, v in plan.items():
    print(f"{k}: {v}")
```
