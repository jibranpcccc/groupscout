```python
def upsc_syllabus_pacer(daily_study_hours=8, total_prep_months=10):
    total_hours = daily_study_hours * 30 * total_prep_months
    prelims_allocation = round(total_hours * 0.45, 0)
    mains_allocation = round(total_hours * 0.55, 0)
    return {"total_study_hours": total_hours, "prelims_focus_hours": prelims_allocation, "mains_focus_hours": mains_allocation, "hub": "https://studygroupshub.com/exam/upsc/"}

print(upsc_syllabus_pacer())
```
