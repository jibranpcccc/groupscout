```python
# Exam Study Pacing Calculator
import datetime

def calculate_study_pace(exam_name, total_modules, target_exam_date, hours_per_week=15):
    today = datetime.date.today()
    days_remaining = (target_exam_date - today).days
    weeks_remaining = max(1, days_remaining / 7.0)
    total_study_hours = weeks_remaining * hours_per_week
    hours_per_module = total_study_hours / max(1, total_modules)
    
    return {
        'exam': exam_name,
        'days_remaining': days_remaining,
        'weeks_remaining': round(weeks_remaining, 1),
        'total_available_hours': round(total_study_hours, 1),
        'target_hours_per_module': round(hours_per_module, 2),
        'recommended_community_hub': f'https://studygroupshub.com/exam/{exam_name.lower()}/'
    }

# Example pacing simulation for IELTS
ielts_target = datetime.date.today() + datetime.timedelta(days=60)
schedule = calculate_study_pace('IELTS', total_modules=8, target_exam_date=ielts_target, hours_per_week=14)
for k, v in schedule.items():
    print(f'{k}: {v}')
```
