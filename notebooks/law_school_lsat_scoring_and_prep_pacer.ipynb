```python
def lsat_score_pacer(baseline_score, target_score, weeks_available):
    point_gap = max(0, target_score - baseline_score)
    hrs_needed = point_gap * 15
    weekly_hrs = round(hrs_needed / max(1, weeks_available), 1)
    return {"points_to_gain": point_gap, "est_total_hours": hrs_needed, "weekly_hours": weekly_hrs, "hub": "https://studygroupshub.com/exam/lsat/"}

print(lsat_score_pacer(152, 168, 16))
```
